```python
import numpy as np
```

```python
# Loss functions

# FOR Linear regression

y_real = np.array([10, 12, 4, 15, 3, 28, 14, 5, 16, 21])
y_pr = np.array([8.5, 12.1, 5, 13.5, 3.4, 25.5, 13.8, 5, 17, 21.3])
# Good prediction - model predicts better than const(mean and median)))
```

```python
# MSE

# MSE = Σ (yi − pi)**2 / n


def mean_squared_error(y_true, y_pred):
    n = len(y_true)
    mse = sum((y_true[i] - y_pred[i])**2 for i in range(n)) / n
    return mse

print(f"MSE: {mean_squared_error(y_real, y_pr)}")
```

```python
# MAE

# MAE = ∑ |yt−ˆyt| / n

def mean_absolute_error(y_true, y_pred):
    n = len(y_true)
    mae = sum(abs(y_true[i] - y_pred[i]) for i in range(n)) / n
    return mae

print(f"MAE: {mean_absolute_error(y_real, y_pr)}")
```

```python
# Binary classifiction

# Accuracy 

y_real = np.array([0, 1, 0, 0, 0, 1, 0, 1, 0, 0])
y_pred = np.array([1, 1, 0, 0, 1, 0, 0, 1, 1, 0])

def accuracy(y_true, y_pred):
    correct = sum(y_true[i] == y_pred[i] for i in range(len(y_true)))
    acc = correct / len(y_true)
    return acc

print(f"Accuracy: {accuracy(y_real, y_pred)}")
```

```python
# F1 score

def f1_score(y_true, y_pred):
    true_positives = sum(y_true[i] == 1 and y_pred[i] == 1 for i in range(len(y_true)))
    false_positives = sum(y_true[i] == 0 and y_pred[i] == 1 for i in range(len(y_true)))
    false_negatives = sum(y_true[i] == 1 and y_pred[i] == 0 for i in range(len(y_true)))

    precision = true_positives / (true_positives + false_positives)
    recall = true_positives / (true_positives + false_negatives)

    f1 = 2 * (precision * recall) / (precision + recall)
    return f1


print(f"F1 score: {f1_score(y_real, y_pred)}")
```

```python
# LINEAR REGRESSION
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

loss_list = []
def linear_regression(X, y, learning_rate=0.0001, epochs=1000):
    n = len(X)
    w0 = 0
    w1 = 0

    for _ in range(epochs):
        y_pred = w0 + w1 * X
        dw0 = (-2 / n) * sum(y - y_pred)
        dw1 = (-2 / n) * sum(X * (y - y_pred))

        w0 -= learning_rate * dw0
        w1 -= learning_rate * dw1

        loss_list.append(mean_squared_error(y, y_pred))

    return w0, w1

```
